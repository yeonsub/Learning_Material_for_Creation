```python
import nltk

# Download the 'punkt' tokenizer models (required for sent_tokenize and word_tokenize)
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

# Download the 'punkt_tab' resource (required for word_tokenize)
try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize

text = "NLTK is a powerful library for natural language processing in Python. It works well in Google Colab!"

# Tokenize the text into words
words = word_tokenize(text)
print(words)

text = "NLTK이 한글도 잘 처리하여 주면 좋겠는데"

# Tokenize the text into words
words = word_tokenize(text)
print(words)

```

```python
from nltk.tokenize import sent_tokenize

text1 = "Apple's name was inspired by Steve Jobs' visit to an apple farm. His visit was while on a fruitarian diet."
text2 = "나는 생성형 AI를 만들 예정이다. 예제 프로그램을 통해 아이디어를 얻고 있다."

# Tokenize the text into sentences
sentence1 = sent_tokenize(text1)
sentence2 = sent_tokenize(text2)

for i, sentence in enumerate(sentence1):
    print(f"Sentence {i+1}: {sentence}")
for i, sentence in enumerate(sentence2):
    print(f"Sentence {i+1}: {sentence}")
```

정규화된 문장만 처리

```python
from nltk.tokenize import RegexpTokenizer

sentence = "NLTK이 한글도 잘 처리하여 주면 좋겠는데 Isn't this great? I can't tell."

# Tokenizer that keeps alphabetic characters (including Unicode), numbers, underscores,
# and apostrophes/hyphens as tokens
tokenizer = RegexpTokenizer(r"[\w'-]+")

tokens = tokenizer.tokenize(sentence)

print(tokens)
```

#한글까지 포함된 embedding

FastText는 다양한 언어에 대한 사전 훈련된 모델을 제공합니다. 한국어 모델은 다음 링크에서 다운로드할 수 있습니다: [FastText Pre-trained Korean Word Vectors](https://fasttext.cc/docs/en/pretrained-vectors.html)

여기서는 `cc.ko.300.vec.gz` 파일을 다운로드하여 사용한다고 가정하고 진행하겠습니다. 이 파일을 Colab 환경에 업로드하거나, `wget` 명령어를 사용하여 직접 다운로드할 수 있습니다.

```python
# cc.ko.300.bin 파일이 이미 다운로드 되어있다고 가정합니다.
# 다음 라인의 자동 다운로드되지 않음. 수동으로 COLAB에 올려야 함
#!wget https://dl.fbaipublicfiles.com/fasttext/vectors-wiki/cc.ko.300.bin

!{sys.executable} -m pip install fasttext
import fasttext
import os

# 모델 파일 경로를 지정합니다. (다운로드된 파일명에 따라 변경될 수 있습니다.)
model_path = 'cc.ko.300.bin'

# 모델 파일이 현재 디렉토리에 없으면 다운로드 시도 (예시)
if not os.path.exists(model_path):
    print(f"{model_path} not found. Attempting to download.")
    try:
        !wget https://dl.fbaipublicfiles.com/fasttext/vectors-wiki/cc.ko.300.bin
    except Exception as e:
        print(f"Error downloading model: {e}")
        print("Please manually download 'cc.ko.300.bin' from https://fasttext.cc/docs/en/pretrained-vectors.html and upload it.")

# 모델 로드
try:
    ko_model = fasttext.load_model(model_path)
    print("Korean FastText model loaded successfully!")
except ValueError as e:
    print(f"Error loading FastText model: {e}")
    print("Please ensure 'cc.ko.300.bin' is correctly downloaded and is not corrupted.")
```

```python
words_to_embed = ["사랑", "인공지능", "대한민국", "apple", "computer"]

for word in words_to_embed:
    if word in ko_model:
        vector = ko_model.get_word_vector(word)
        print(f"\nEmbedding for '{word}':")
        print(vector[:10]) # 첫 10개 요소만 출력하여 너무 길어지지 않게 합니다.
        print(f"Embedding dimension for '{word}': {len(vector)}")
    else:
        print(f"\nWord '{word}' not found in vocabulary.")
```

#영어 Embedding

```python
words_to_process = ["사랑", "love", "you", "korea"]

for word in words_to_process:
    if word in glove_model:
        word_vector = glove_model[word]
        print(f"\nEmbedding for '{word}':", word_vector)
        print(f"Embedding dimension for '{word}':", len(word_vector))
    else:
        print(f"Word '{word}' not found in vocabulary.")
```
